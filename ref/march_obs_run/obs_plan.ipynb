```python
def make_vis(coords,name):
    coords = coords.split()
    ra = coords[0]
    dec = coords[1]
    
    vis = astroplan.FixedTarget(coord=SkyCoord(ra, dec, unit=(u.hourangle, u.degree)),
                                   name=name)
    return vis
```

```python
def parse_tab_str(tab_str):
    rows = []
    for line in tab_str.strip().split('\n'):
        parts = line.split()
        name = parts[0]
        ra, dec = parts[1], parts[2]
        mag = float(parts[3])
        nexp, exptime = int(parts[4].split('x')[0]), float(parts[4].split('x')[1])
        coord = SkyCoord(ra, dec, unit=(u.hourangle, u.degree))
        rows.append({'name': name, 'coord': coord, 'mag': mag, 'nexp': nexp, 'exptime': exptime})
    return rows

# --- Catalog file: all_tab sorted by RA ---
all_rows = parse_tab_str(all_tab)
all_rows_sorted = sorted(all_rows, key=lambda r: r['coord'].ra.hour)
dir = "/Users/yize/Library/CloudStorage/Dropbox/research/TNS/observing/Gagliano_Magellan_0321_2026/ObsPlan/"
with open(dir + "Gagliano_LDSS_0321_catalog", "w") as f:
    for i, row in enumerate(all_rows_sorted, 1):
        ra_str  = row['coord'].ra.to_string(unit=u.hourangle, sep=':', precision=2, pad=True)
        dec_str = row['coord'].dec.to_string(sep=':', precision=2, alwayssign=True, pad=True)
        f.write(f"{i}\t{row['name']}\t{ra_str}\t{dec_str}\t2000.0\t0.0\t0.0\t-62.5\tHRZ\t"
                f"00:00:00.0\t+00:00:00\t2000.0\t00:00:00.0\t+00:00:00\t2000.0\n")
print("Wrote catalog")
cat_index = {row['name']: i+1 for i, row in enumerate(all_rows_sorted)}

# --- Timeline file: tab in order with computed UT times ---
tab_rows = parse_tab_str(tab)
readout = 100   # seconds
slew    = 300  # seconds
t = Time((pre_twi15deg + 6 / 60 / 24).iso)
start_times, end_times = [], []
for row in tab_rows:
    obs_dur = row['nexp'] * row['exptime'] + (row['nexp'] - 1) * readout
    end_t = t + obs_dur * u.second
    start_times.append(t)
    end_times.append(end_t)
    t = end_t + slew * u.second

standards = {'LTT2415', 'LTT6248', 'LTT9239', 'LTT1020', 'LTT4364', 'LTT4816', 'LTT1788'}

with open(dir + "Gagliano_LDSS_0321_timeline", "w") as f:
    f.write("#  Target     UT          Observation     Comments\n")
    for i, (row, t_start, t_end) in enumerate(zip(tab_rows, start_times, end_times)):
        idx     = cat_index.get(row['name'], '?')
        obs_str = f"spec: {row['nexp']}x{int(row['exptime'])}s"
        comment = f"  {row['mag']} mag star" if row['name'] in standards else ""

        if i == 0:
            # First target: "before <start of next>"
            next_start = start_times[i + 1].iso[11:16]
            time_str = f"before {next_start}"
        elif i == len(tab_rows) - 1 and row['name'] in standards:
            # Last target is a standard: "after <its start time>"
            time_str = f"after {t_start.iso[11:16]}"
        else:
            # Middle targets: end = start of next (slew included in window)
            disp_end = start_times[i + 1].iso[11:16] if i < len(tab_rows) - 1 else t_end.iso[11:16]
            time_str = f"{t_start.iso[11:16]} - {disp_end}"

        f.write(f"{idx} {row['name']} {time_str} {obs_str}{comment}\n")
print("Wrote timeline")
```
